# DUPLICATE ROWS

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'A', 'B'],
    'Age': [20, 25, 30, 20, 25]
})

print("Dataset:")
print(df)


print("\nDuplicate Rows:")
print(df.duplicated())


df_clean = df.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(df_clean)
```
