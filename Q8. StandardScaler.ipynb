# StandardScaler

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

df = pd.DataFrame({
    'Age': [20, 25, 30, 35, 40],
    'Salary': [20000, 25000, 30000, 35000, 40000]
})

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

print("Scaled Data:")
print(scaled_data)
```
